```python

%sql
-- Create storage credential and external location (run once)
CREATE EXTERNAL LOCATION IF NOT EXISTS taxi_location URL
'abfss://taxi@siadp203sa.dfs.core.windows.net/' WITH (STORAGE CREDENTIAL sia_dp203_databricks);

CREATE SCHEMA IF NOT EXISTS test.raw
COMMENT 'Camada crua dos dados de táxi';
```

```python

%sql
-- Create volume for raw data

CREATE EXTERNAL VOLUME IF NOT EXISTS test.raw.taxi
LOCATION 'abfss://taxi@siadp203sa.dfs.core.windows.net/raw'
COMMENT 'Volume for raw taxi data files from Azure storage';
```

```python
%sql
-- Criação do catálogo
CREATE CATALOG IF NOT EXISTS test
MANAGED LOCATION 'abfss://taxi@siadp203sa.dfs.core.windows.net/'
COMMENT 'Catálogo de dados de corridas de táxi de Nova York no Databricks com Delta Lake de teste';
```

```python

%sql
-- Schemas (camadas do lakehouse)
CREATE SCHEMA IF NOT EXISTS test.bronze
COMMENT 'Camada bruta dos dados de táxi';

CREATE SCHEMA IF NOT EXISTS test.silver
COMMENT 'Camada tratada e enriquecida dos dados de táxi';

CREATE SCHEMA IF NOT EXISTS test.gold
COMMENT 'Camada agregada dos dados de táxi para consumo analítico';
```
